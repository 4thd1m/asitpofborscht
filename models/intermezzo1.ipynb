```python
import torch
import matplotlib.pyplot as plt
%matplotlib inline
plt.rcParams["figure.figsize"] = (4, 3)

def f(x: torch.Tensor):
  return x ** 2

x = torch.linspace(-3, 3, 200)
y = f(x)
plt.plot(x, y)
```

```python
import torch
import matplotlib.pyplot as plt
%matplotlib inline
plt.rcParams["figure.figsize"] = (4, 3)

def f(x: torch.Tensor):
  return x.sqrt()

x = torch.linspace(-3, 3, 200)
y = f(x)
plt.plot(x, y)
plt.plot(x, -y)
```

```python
import torch
import matplotlib.pyplot as plt
%matplotlib inline
plt.rcParams["figure.figsize"] = (4, 3)

def f(x: torch.Tensor):
  return x.sqrt()

def g(x: torch.Tensor):
  return x.sqrt()

x = torch.linspace(-3, 3, 200)
y = g(f(x))
plt.plot(x, y)
```

```python
import torch
import matplotlib.pyplot as plt
%matplotlib inline
plt.rcParams["figure.figsize"] = (4, 3)

def f(x: torch.Tensor):
  return x.abs()

x = torch.linspace(-3, 3, 200)
y = f(x)
plt.plot(x, y)
```
